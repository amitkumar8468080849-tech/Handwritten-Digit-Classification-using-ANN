```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Flatten
```

```python
(X_train,y_train),(X_test,y_test)=keras.datasets.mnist.load_data()
```

```python
X_train=X_train/255
X_test=X_test/255
```

```python
X_train#all values range(0,1)
```

```python
X_test
```

```python
model=Sequential()
model.add(Flatten(input_shape=(28,28)))#data ko 1D me convetion
model.add(Dense(128,activation='relu'))
model.add(Dense(32,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

```python
#spparse_categorical_crossentropyka use karane per one hot encooding ka ues nahi karana padata hai
model.compile(loss='sparse_categorical_crossentropy',optimizer='Adam',metrics=['accuracy'])
```

```python
bistory=model.fit(X_train,y_train,epochs=100,validation_split=0.2)
```

```python
y_prob=model.predict(X_test)
y_prob
```

```python
y_pred=y_prob.argmax(axis=1)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
plt.plot(bistory.history['loss'])
plt.plot(bistory.history['val_loss'])
```

```python
plt.plot(bistory.history['accuracy'])
plt.plot(bistory.history['val_accuracy'])
```
